3. Convert temperature from Celsius to Fahrenheit.


```python
# formula: °F = (°C × 9/5) + 32

temp_c = float(input("Enter temperature in Celsius: "))

temp_f = (temp_c * 9/5) + 32

print("Temperature in Fahrenheit:", temp_f)
```

4. Write a Python program to calculate the area of a rectangel given its length and width.

```python
length = float(input("Enter length of rectangle: "))
width = float(input("Enter width of rectangle: "))

area = length * width

print("Area of rectangle:", round(area,2))
```

7. Given a list of numbers, find the maximum and minimum values.

```python
lst = [2, 7, 4, 9, 5, 1, 6, 3, 2]

small = None
large = None
for num in lst:
    if small == None or small > num:
        small = num

    if large == None or large < num:
        large = num

print(f"Largest number is {large} and smallest number is {small}.")
```

8. Create a Python function to check if a given string is a palindrome.

```python
line = input("Enter a string")
text = line.replace(" ","")

for i in range(len(text)):
    if text[i] != text[-i]: continue

if i == len(text)-1: 
    print("Given string is a palindrome")    
```

1. Calculate the compound interest for a given principal amount, interest rate, and time period.

```python
pr = float(input("Enter principal amount: "))
ir = float(input("Enter interest rate: "))
tp = float(input("Enter time period (years): "))

ci = pr * (1 + ir/1)^tp

print("Compound interest is",ci)
```
