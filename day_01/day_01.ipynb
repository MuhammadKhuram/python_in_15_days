**3. Convert temperature from Celsius to Fahrenheit.**


```python
# formula: °F = (°C × 9/5) + 32

temp_c = float(input("Enter temperature in Celsius: "))

temp_f = (temp_c * 9/5) + 32

print("Temperature in Fahrenheit:", temp_f)
```

**7. Given a list of numbers, find the maximum and minimum values.**

```python
lst = [2, 7, 4, 9, 5, 1, 6, 3, 2]

small = None
large = None
for num in lst:
    if small is None or small > num:
        small = num

    if large is None or large < num:
        large = num

print(f"Largest number is {large} and smallest number is {small}.")
```

**9. Calculate the compound interest for a given principal amount, interest rate, and time period.**

```python
pa = float(input("Enter principal amount: "))
ir = float(input("Enter interest rate (in decimals. e.g. 3% = 0.03): "))
tp = float(input("Enter time period (years): "))

ci = pa * (1 + ir)**tp - pa

print("Compound interest is",round(ci,3))
```

**11. Given a list of integers, find sum of all positive numbers.**

```python
lst = [1, 3, 0, -3, 5, -1, -4, 7, 13, -28]
total = 0

for i in lst:
    if i > 0:
        total += i

print(f"Sum of all positive numbers is {total}.")
```
