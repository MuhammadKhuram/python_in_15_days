**6. Implement a program that checks if a given string is a palindrome.**

```python
text = input("Enter a string: ").replace(" ","")

text_rev = str()

for i in range(len(text)):
    text_rev += text[-i-1]

if text_rev == text:
    print("Given string is a palindrome.")
else:
    print("Given string is not a palindrome.")
```

**7. Create a function to reversea given string.**

```python
def string_reverse(x):
    y = ""
    for i in range(len(x)):
        y += x[-i-1]
    return y

text = input("Enter a string: ")
text_reverse = string_reverse(text)

print(f"Orignal String: {text}")
print(f"Reversed String: {text_reverse}")
```

**9. Write a Python program to check if a given string is a pangram ( contain all letters of the alphabet).**

```python
# e.g.
# Pack my box with five dozen liquor jugs
#The quick brown fox jumps over the lazy dog

def is_pangram(text):
    letters = ""
    for char in text.lower():
        if char.isalpha() and char not in letters:
            letters += char

    return len(letters)

if is_pangram(input("enter a string: ")) == 26:
    print("Given string is a pangram.")
else:
    print("Given string is not a pangram.")
```

**Write a program to check if a number is a prime.**

```python
import math as m

def check_prime(x):
    sqroot = round(m.sqrt(x))
    for i in range(2,sqroot+1):
        if x % i != 0: 
            continue
        else: 
            return False

    return True

num = int(input("Enter a number: "))

if check_prime(num):
    print(f"{num} is a prime number")

else:
    print(f"{num} is not a prime number")
```
